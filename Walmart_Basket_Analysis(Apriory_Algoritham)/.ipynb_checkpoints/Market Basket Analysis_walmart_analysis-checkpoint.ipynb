<h1 style="text-align: center;">Load Data & PostProcessing </h1>

```
one basket --> one Invoice --> one transaction

-ve Quantity means Return Items, we need to remove them
Remove rows where InvoiceNO is Null
Romve rows where Invoice contains 'C', that is acutally the invoice

Description is the Product Name
```

````python
df = pd.read_csv('data/Online_Retail.csv')
print(df.shape)
df.head(2)
````

````python
# Remove Noice
df = df[df.Quantity > 0]                   # Remove returned items
df = df.dropna(subset=['InvoiceNo'])
df = df[~df.InvoiceNo.str.contains('C')]   # Remove all the transaction done by credit card

# Only Take United Kingdom Data
df = df[df.Country == 'United Kingdom']

# Taking only requred columns for Basket-Item Analytisis
df = df[['InvoiceNo','Description','Quantity']].rename(columns={'InvoiceNo':'Basket','Description':'Item'})
df.head()
````

````python
print(f"No of Baskets we have: {df.Basket.nunique()}")  # 18786
print(f"No of Items we have : {df.Item.nunique()}")     # 4058
````

## Now creating Bucket-Item  (18786 * 4058) One Hot Encoded Matrixs

````python
# Create Bucket(Transaction) Item(Description) Matrics (Basket Item Occurance Metrics)
df_bucket_item = df.groupby(['Basket','Item'])['Quantity'].sum().unstack().fillna(0)
print(df_bucket_item.shape)
df_bucket_item.head()
````

````python
# Make Data Binary
df_bucket_item = df_bucket_item.applymap(lambda x: 0 if x == 0 else 1)
````

## Filtering out Baskets based on basket size  : There should be atleast 2 items in the basket

````python
# Take only thoes Invoices in which we have atleast 2 item (Bucket Needs to be have atleast 2 items)
df_bucket_item = df_bucket_item[df_bucket_item.sum(axis=1)>=2]
print(df_bucket_item.shape)
df_bucket_item.head(2)
````

<h1 style="text-align: center;">Apriory Algorithaham </h1>

````python
from mlxtend.frequent_patterns import apriori,association_rules
````

```
Drawback : If item pool is very large with size n , then total possible combination is also very large (2^n)

Now to reduce complexity, idea, is lets say 
if frequency of item {a} in baskets is 20 then the frequency of {a,b} will be less the {a} in the baskets
so if item a occured x no of times in out of n baskets then any combination of item a with other item 
will have frequency less than or equal to x
```

```
creating hurestic rule if frequency of any combination is less than threshold 
then we will stop including that conbination
```

```
This is done by using parameter called support
```
$$
Support = \frac{Occurrence\ of\ the\ item\ combination\ in\ baskets}{Total\ number\ of\ baskets}
$$

$$
Support \in [0,1]
$$

```
                        suport = 1   : Item Combination Occured in all the baskets
                               = 0   : Item Combination Occured in None of the baskets
So Single Items will have Usually highter support value
As the Query Item Combination size increases Support will be deacreses 
```

### Finding Support Of Each Item Combination by placing min threshold on support by providing Basket-Item Matrics

````python
most_items_occured_togeter = apriori(df_bucket_item,
                                     min_support=0.03,                         # setting threshold on support
                                     use_colnames=True
                                    ).sort_values('support',
                                                  ascending=False
                                                 ).reset_index(drop=True)

most_items_occured_togeter.head()
````

````python
# Remvoving single Item Combination
most_items_occured_togeter['len_of_combination'] = most_items_occured_togeter.itemsets.agg(len)
most_items_occured_togeter_pair= most_items_occured_togeter[most_items_occured_togeter.len_of_combination >1].reset_index(drop=True)
most_items_occured_togeter_pair.head()
````

## Association Rules Mining for Binary Item Combination Recommendation assuming we know the first Item (Anticedent)

```
P(Coke/Pizza), P(Bread/Milk) , P(Butter/Bread)is High 
Pizza,Milk,Bread  : Anticedent
Coke,Bread,Butter : Consequent

Note : P(Coke/Pizza) != P(Pizza/Coke)  i.e Order of the association is Important
```

### Association Metrics

#### Confidence

$$
\text{Confidence(Coke/Pizza)} 
= \frac{\text{Support(Coke, Pizza)}}{\text{Support(Pizza)}}
= \frac{\text{Number of baskets having Coke and Pizza together}}
{\text{Number of baskets having Pizza}}
\ \in [0, 1]
$$

#### Lift

$$
\text{Lift(Coke, Pizza)} 
= \frac{\text{Support(Coke, Pizza)}}
{\text{Support(Coke)} \times \text{Support(Pizza)}}
\ \in [0, \infty)
$$

```
                lift(X,Y) = 1  : if x and y have no correlation then support(x,y) = support(x) * support(y)
                          > 1  : if x and y have positive correlation
                          < 1  : if x and y have negative correlation
```

#### Leverage

$$
\text{Leverage(A,B)} 
= \text{Support(A, B)} 
- \text{Support(A)} \times \text{Support(B)}
$$

#### Conviction

$$
\text{Conviction(A,B)} 
= \frac{1 - \text{Support(B)}}
{1 - \text{Confidence(A → B)}}
\ \in [-1,1]
$$

```
                    Conviction(A,B) = 0 : no correlation
                                    > 0 : possitiove correlation between A and B
                                    < 0 : negative correlation between A and 
```

````python
associtaion_rules_df = association_rules(most_items_occured_togeter,
                                         metric='lift',
                                         min_threshold=1        # To see only positive association
                                        ).sort_values('lift',
                                                      ascending=False
                                                     ).reset_index(drop=True)
associtaion_rules_df
````
